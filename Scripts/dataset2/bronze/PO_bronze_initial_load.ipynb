```python
# Welcome to your new notebook
# Type here in the cell editor to add cod
from datetime import datetime
from pyspark.sql.functions import col
from pyspark.sql.functions import lit
```

```python
now=str(datetime.utcnow())
now
```

```python
df=spark.table("staging.production_orders")
```

```python
df=df.withColumn("valid_from",lit(now))
```

```python
df=df.withColumn("valid_from",col("valid_from").cast("timestamp"))
```

```python
df=df.withColumn("valid_to",lit(None))
```

```python
df=df.withColumn("valid_to",col("valid_to").cast("timestamp"))
```

```python
display(df)
```

```python
df.write.mode("overwrite").saveAsTable("bronze.production_orders")
```
