```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
df=spark.sql("""
SELECT *
FROM silver.production_orders
WHERE Nid not LIKE '%WO%'
""")
```

```python
df.createOrReplaceTempView("v_df")
```

```python
v_df=spark.sql("select * from v_df")
```

```python
cols = [c for c in v_df.columns]

update_set = ",\n".join([f"a.{c} = b.{c}" for c in cols if c != "NId"])
insert_cols = ", ".join(cols)
insert_vals = ", ".join([f"b.{c}" for c in cols])
```

```python
spark.sql(f"""
MERGE INTO gold.production_orders a
USING v_df b
ON a.NId=b.NId
WHEN MATCHED THEN UPDATE SET {update_set}
WHEN NOT MATCHED THEN INSERT ({insert_cols})
VALUES ({insert_vals})
""")
```
